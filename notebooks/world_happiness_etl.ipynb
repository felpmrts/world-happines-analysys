# 🌍 World Happiness Report - ETL Pipeline 🛠️

This notebook documents the **Extract, Transform, and Load (ETL)** process for the World Happiness datasets (2015-2019). The goal is to clean and consolidate yearly data into a single "Golden Table" for analysis. 🚀

---

## 1. Initial Setup ⚙️

First, we import the `pandas` library for data manipulation and define the folder structure. Keeping `raw` and `processed` data separate is a best practice in Data Engineering to ensure data lineage. 📂

```python
# Imports & paths

import pandas as pd

# Data paths
raw_path = '../data/raw/'
processed = '../data/processed/'
```

## 2. Extract Phase 📥

The data is spread across five CSV files. During extraction, we manually inject a `Year` column into each DataFrame. This is essential because once we merge them, we need to distinguish which year each record belongs to. 📑

```python
# Extract

df_2015 = pd.read_csv(f"{raw_path}happiness_2015.csv")
df_2015["Year"] = 2015

df_2016 = pd.read_csv(f"{raw_path}happiness_2016.csv")
df_2016["Year"] = 2016

df_2017 = pd.read_csv(f"{raw_path}happiness_2017.csv")
df_2017["Year"] = 2017

df_2018 = pd.read_csv(f"{raw_path}happiness_2018.csv")
df_2018["Year"] = 2018

df_2019 = pd.read_csv(f"{raw_path}happiness_2019.csv")
df_2019["Year"] = 2019
```

## 3. Transform Phase 🧪

This is where the "heavy lifting" happens. Since column names changed every year (e.g., `Happiness Rank` vs `Overall rank`), we must standardize them to a single schema. 🧹

### Key Actions:
* **Standardization:** Renaming columns to a unified format (e.g., `Economy (GDP)`, `Health`). 📋
* **Feature Selection:** Keeping only the most relevant columns for analysis. ✅
* **Data Cleaning:** Handling missing values in the 2018 dataset using `.fillna(0)`. 🧼

```python
# Tranform

new_df_2015 = df_2015.copy()
new_df_2016 = df_2016.copy()
new_df_2017 = df_2017.copy()
new_df_2018 = df_2018.copy()
new_df_2019 = df_2019.copy()

# Rename

final_columns = ["Year","Rank", "Country", "Happiness Score", "Economy (GDP)", "Freedom", "Trust (Government)", "Health"]

new_df_2015 = new_df_2015.rename(columns={
    "Happiness Rank": "Rank",
    "Country": "Country",
    "Happiness Score": "Happiness Score",
    "Economy (GDP per Capita)": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust (Government Corruption)": "Trust (Government)",
    "Health (Life Expectancy)": "Health"
})[final_columns]

new_df_2016 = new_df_2016.rename(columns={
    "Happiness Rank": "Rank",
    "Country": "Country",
    "Happiness Score": "Happiness Score",
    "Economy (GDP per Capita)": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust (Government Corruption)": "Trust (Government)",
    "Health (Life Expectancy)": "Health"
})[final_columns]

new_df_2017 = new_df_2017.rename(columns={
    "Happiness.Rank": "Rank",
    "Country": "Country",
    "Happiness.Score": "Happiness Score",
    "Economy..GDP.per.Capita.": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust..Government.Corruption.": "Trust (Government)",
    "Health..Life.Expectancy.": "Health"
})[final_columns]

new_df_2018 = new_df_2018.rename(columns={
    "Overall rank": "Rank",
    "Country or region": "Country",
    "Score": "Happiness Score",
    "GDP per capita": "Economy (GDP)",
    "Freedom to make life choices": "Freedom",
    "Perceptions of corruption": "Trust (Government)",
    "Healthy life expectancy": "Health"
})[final_columns]

new_df_2019 = new_df_2019.rename(columns={
    "Overall rank": "Rank",
    "Country or region": "Country",
    "Score": "Happiness Score",
    "GDP per capita": "Economy (GDP)",
    "Freedom to make life choices": "Freedom",
    "Perceptions of corruption": "Trust (Government)",
    "Healthy life expectancy": "Health"
})[final_columns]

# Data processing

new_df_2018["Trust (Government)"] = new_df_2018["Trust (Government)"].fillna(0)
```

## 4. Load Phase 💾

In the final stage, we stack all yearly DataFrames vertically using `pd.concat`. We use `ignore_index=True` to create a fresh index, avoiding duplicates. Finally, we export the result to the `processed` folder. 🔗

```python
# Load

df_final = pd.concat([new_df_2015, new_df_2016, new_df_2017, new_df_2018, new_df_2019], ignore_index=True)

df_final.to_csv(f'{processed}happiness_complete.csv', index=False)

print("LOAD sucess!")
```
