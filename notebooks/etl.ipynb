```python
# Imports & paths

import pandas as pd

# Data paths
raw_path = '../data/raw/'
processed = '../data/processed/'
```

```python
# Extract

df_2015 = pd.read_csv(f"{raw_path}happiness_2015.csv")
df_2015["Year"] = 2015

df_2016 = pd.read_csv(f"{raw_path}happiness_2016.csv")
df_2016["Year"] = 2016

df_2017 = pd.read_csv(f"{raw_path}happiness_2017.csv")
df_2017["Year"] = 2017

df_2018 = pd.read_csv(f"{raw_path}happiness_2018.csv")
df_2018["Year"] = 2018

df_2019 = pd.read_csv(f"{raw_path}happiness_2019.csv")
df_2019["Year"] = 2019
```

```python
# Info & Exploration

print(f"--- INFO 2015 ---\n")
df_2015.info()

print(f"\n--- INFO 2016 ---\n")
df_2016.info()

print(f"\n--- INFO 2017 ---\n")
df_2017.info()

print(f"\n--- INFO 2018 ---\n")
df_2018.info()

print(f"\n--- INFO 2019 ---\n")
df_2019.info()
```

```python
# Tranform

new_df_2015 = df_2015.copy()
new_df_2016 = df_2016.copy()
new_df_2017 = df_2017.copy()
new_df_2018 = df_2018.copy()
new_df_2019 = df_2019.copy()

# Rename

final_columns = ["Year","Rank", "Country", "Happiness Score", "Economy (GDP)", "Freedom", "Trust (Government)", "Health"]

new_df_2015 = new_df_2015.rename(columns={
    "Happiness Rank": "Rank",
    "Country": "Country",
    "Happiness Score": "Happiness Score",
    "Economy (GDP per Capita)": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust (Government Corruption)": "Trust (Government)",
    "Health (Life Expectancy)": "Health"
})[final_columns]

new_df_2016 = new_df_2016.rename(columns={
    "Happiness Rank": "Rank",
    "Country": "Country",
    "Happiness Score": "Happiness Score",
    "Economy (GDP per Capita)": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust (Government Corruption)": "Trust (Government)",
    "Health (Life Expectancy)": "Health"
})[final_columns]

new_df_2017 = new_df_2017.rename(columns={
    "Happiness.Rank": "Rank",
    "Country": "Country",
    "Happiness.Score": "Happiness Score",
    "Economy..GDP.per.Capita.": "Economy (GDP)",
    "Freedom": "Freedom",
    "Trust..Government.Corruption.": "Trust (Government)",
    "Health..Life.Expectancy.": "Health"
})[final_columns]

new_df_2018 = new_df_2018.rename(columns={
    "Overall rank": "Rank",
    "Country or region": "Country",
    "Score": "Happiness Score",
    "GDP per capita": "Economy (GDP)",
    "Freedom to make life choices": "Freedom",
    "Perceptions of corruption": "Trust (Government)",
    "Healthy life expectancy": "Health"
})[final_columns]

new_df_2019 = new_df_2019.rename(columns={
    "Overall rank": "Rank",
    "Country or region": "Country",
    "Score": "Happiness Score",
    "GDP per capita": "Economy (GDP)",
    "Freedom to make life choices": "Freedom",
    "Perceptions of corruption": "Trust (Government)",
    "Healthy life expectancy": "Health"
})[final_columns]

# Data processing

new_df_2018["Trust (Government)"] = new_df_2018["Trust (Government)"].fillna(0)
```

```python
# Load

df_final = pd.concat([new_df_2015, new_df_2016, new_df_2017, new_df_2018, new_df_2019], ignore_index=True)

df_final.to_csv(f'{processed}happiness_complete.csv', index=False)

print("Etapa de LOAD finalizada com sucesso!")
```
