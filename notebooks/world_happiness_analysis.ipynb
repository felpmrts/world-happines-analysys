```python
# Imports & paths

import pandas as pd
import numpy as np

# Paths

processed = '../data/processed/happiness_complete.csv'

# Define

pd.set_option('display.max_rows', 1000)
pd.set_option('display.precision', 2)
```

```python
# Extract & Exploration

df = pd.read_csv(processed, low_memory=False)
df_copy = df.copy()
```

## Analysis

```python
# Separating the years

df_2015 = df_copy[df_copy['Year'] == 2015]
df_2016 = df_copy[df_copy['Year'] == 2016]
df_2017 = df_copy[df_copy['Year'] == 2017]
df_2018 = df_copy[df_copy['Year'] == 2018]
df_2019 = df_copy[df_copy['Year'] == 2019]
```

## Top 10

```python
# Top 10 most happiness countries by years

print(f"\nTop 10 most happiness countries by years")
most_happines = df_copy.groupby('Country')['Happiness Score'].mean().sort_values(ascending=False).head(10)
display(most_happines.to_frame().style.format(precision=2)) 

# Top 10 less happiness countries by years

print(f"\nTop 10 less happiness countries by years")
less_happines = df_copy.groupby('Country')['Happiness Score'].mean().sort_values(ascending=True).head(10)
display(less_happines.to_frame().style.format(precision=2)) 
```

## Wealth vs Happiness

```python
# Wealth Vs Happiness

print(f"\n10 most")
happinessXeconomy = df_copy.groupby('Happiness Score')['Economy (GDP)'].mean().sort_values(ascending=False)
display(happinessXeconomy.to_frame().head(10))

print(f"\n10 less")
happinessXeconomy = df_copy.groupby('Happiness Score')['Economy (GDP)'].mean().sort_values(ascending=True)
display(happinessXeconomy.to_frame().head(10))
```

## Health


```python
correlation = df_copy[['Happiness Score', 'Economy (GDP)', 'Health']].corr()
display(correlation.style.background_gradient(cmap='coolwarm').format(precision=2))
```

## Growth Happiness 

```python
total_happines2015 = (df_2015['Happiness Score'].sum())/158
total_happines2016 = (df_2016['Happiness Score'].sum())/157
total_happines2017 = (df_2017['Happiness Score'].sum())/155
total_happines2018 = (df_2018['Happiness Score'].sum())/156
total_happines2019 = (df_2019['Happiness Score'].sum())/156

print(f"\n{total_happines2015}")
print(f"\n{total_happines2016}")
print(f"\n{total_happines2017}")
print(f"\n{total_happines2018}")
print(f"\n{total_happines2019}")
```
